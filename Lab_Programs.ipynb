```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats 

np.random.seed(42)

n, p=20, 0.35
binom = stats.binom(n,p)
x = np.arange(0, n+1)

print(f' Mean = {binom.mean() :.4f} (Theory: np=(n*P))')
```
