```python
import numpy as np
roll=np.random.choice([1,2,3,4,5,6])
print("you rolled roll",roll)

rolls=np.random.choice([1,2,3,4,5,6],size=10)
print("10 rolled",rolls)

rolls=np.random.choice([1,2,3,4,5,6],size=100)
print("100 rolled",rolls)

for face in range(1,7):
  count=np.sum(rolls==face)
  print(face,"was rolled",count,"times","probability",count/1000)
```

```python
rolls=np.random.randint(1,7,10)
for number in range(1,7):
  count=np.sum(rolls==number)
  print(number,"was called",count,"times","probability",count/100)
```
